## Summary of Your Efforts Today

You may have started this workshop with little or no experience working programmatically with LLMs. Over the course of the day, you built a strong foundation across several key areas:

- Sending prompts and receiving responses using both the OpenAI API and LangChain
- Streaming outputs and processing inputs in batches
- Iteratively refining prompts to improve model behavior
- Using prompt templates to create reusable, structured prompts
- Building modular, composable workflows with LangChain Expression Language (LCEL)
- Creating custom runnables and composing chains, including parallel execution
- Working with chat message types (system, human, AI) to control model behavior
- Applying techniques like few-shot prompting and chain-of-thought prompting
- Managing conversation history to build chatbot-style interactions
- Generating and validating structured outputs for use in downstream code
- Performing document analysis and tagging using structured data
- Extending LLM capabilities with tools and building agents that decide when to use them

That's a lot, and again, congratulations! We wouldn't expect anyone to feel fluent in all the topics we covered today after going through the materials just once, so we want to invite and encourage you to return to these materials at your leisure.

## Next Steps

Today's workshop is only one of many DLI offerings on the subject of generative AI and LLMs. We encourage you to check out [the DLI learning path for Generative AI and Large Language Models](https://nvdam.widen.net/s/brxsxxtskb/dli-learning-journey-2009000-r5-web) and continue to grow your capabilities with us.
